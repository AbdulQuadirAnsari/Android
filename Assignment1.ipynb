```python
#1. Given a number, find the sum of its digits.
# Take the number as an input from the user.

num = input("please enter a number ")
numlist = list(num)
sum = 0
for dig in numlist:
    sum+=int(dig)

print(f"sum of its digit is {sum}")
```

```python
# 2. Given a number, check whether the given number is an Armstrong number or not.
# A positive integer is called an Armstrong number of order n if:
# abcd... = an + bn + cn + dn + ...
# Example: 153 = 1*1*1 + 5*5*5 + 3*3*3
# 153 is an Armstrong number of order 3.
# Inputs from the user will be number and order n.

numAndOrder = input("please enter a number and order ")
numAndOrderList = numAndOrder.split(" ")
num = numAndOrderList[0].strip()
order = numAndOrderList[1].strip()

numList = list(num)
sum = 0
for dig in numList:
    sum+=int(dig)**int(order)

if sum == int(num):
    print(f"{num} is an Armstrong number of order {order}")
```

```python
# 3. Given a string, write a python function to check if it is  palindrome or not.
# A string is said to be palindrome if the reverse of the string is the same as string.
# For example, “malayalam” is a palindrome, but “music” is not a palindrome.
str = input("please enter a string").strip().lower()

strlist = list(str)
strlist.reverse()
reversedstr = "".join(strlist)

if str == reversedstr:
  print("your string is a palindrome")
else:
  print("your string is not a palindrome")
```

```python
# A hospital is planning to digitize its patient record system to improve efficiency in managing patient information and billing.
# As a Python developer, you are tasked with building a simple system to process and analyze patient data.
# You are given a collection of patient details, where each record contains the patient’s name, age, disease, and total bill amount.
# Store and organize patient data using appropriate data structures
# Calculate the number of patients affected by each disease
# Identify the patient with the highest bill amount
# Compute the average bill and display patients whose bill exceeds the average
# Ensure that duplicate patient entries are removed using suitable techniques

print('''
Welcome to Ansari Hospital.
''')
patient = list()
record = {}


while(True):
  #name, age, disease, and total bill amount
  name = input("Name: ")
  age  = int(input("Age : ") or "0")
  disease = input("Disease: ")
  totalBillAmount = int(input("Total Bill Amount: ") or "0")

  thisdict = dict(name = name, age = age, disease = disease, totalBillAmount = totalBillAmount)


  patient.append(thisdict)
  thisdict = None
  yesOrNo = input("want to continue? Y/N: ")

  if yesOrNo.lower() == 'n':
    print("\n\n\n");
    break
  else:
    print("\n\n\n");
    continue

uniquePatientList = []
namesSet = set()

for item in patient:
  if item["name"] not in namesSet:
    uniquePatientList.append(item)
    namesSet.add(item["name"])

print("list of patient without duplicate is as follows")
print("=============================================")
print(uniquePatientList)
print("=============================================")
# print("printing list of records")
# print("================================")
# print(patient)

print("\n\nPatients affected by each disease are as follows: ")
print("=============================================\n")
listOfDisease = [item["disease"] for item in uniquePatientList ]
setOfDisease = set(listOfDisease)

for disease in setOfDisease:
  print(f"disease: {disease}, patient affected: {listOfDisease.count(disease)}")

print("\n=============================================\n\n")

patientWithHighestBillAmount = {
    "name":"",
    "totalBillAmount":0
}

patientWithHighestBillAmount = max(uniquePatientList, key = lambda record: record["totalBillAmount"])

print(f"the patient with highest bill amount is {patientWithHighestBillAmount["name"]}")

billAmounts = [item["totalBillAmount"] for item in uniquePatientList]

avgBillAmount = sum(billAmounts)/len(billAmounts)
print(f"the average bill amount is {avgBillAmount}")

namesList = [item["name"] for item in uniquePatientList if item["totalBillAmount"] > avgBillAmount]

print("below are the patients whose bill amounts greater than average")
print("==============================================================")
print(namesList)


```
