```python
#1. Given a number, find the sum of its digits.
# Take the number as an input from the user.

num = input("please enter a number ")
numlist = list(num)
sum = 0
for dig in numlist:
    sum+=int(dig)

print(f"sum of its digit is {sum}")
```

```python
# 2. Given a number, check whether the given number is an Armstrong number or not.
# A positive integer is called an Armstrong number of order n if:
# abcd... = an + bn + cn + dn + ...
# Example: 153 = 1*1*1 + 5*5*5 + 3*3*3
# 153 is an Armstrong number of order 3.
# Inputs from the user will be number and order n.

numAndOrder = input("please enter a number and order ")
numAndOrderList = numAndOrder.split(" ")
num = numAndOrderList[0].strip()
order = numAndOrderList[1].strip()

numList = list(num)
sum = 0
for dig in numList:
    sum+=int(dig)**int(order)

if sum == int(num):
    print(f"{num} is an Armstrong number of order {order}")
```

```python
# 3. Given a string, write a python function to check if it is  palindrome or not.
# A string is said to be palindrome if the reverse of the string is the same as string.
# For example, “malayalam” is a palindrome, but “music” is not a palindrome.
str = input("please enter a string").strip().lower()

strlist = list(str)
strlist.reverse()
reversedstr = "".join(strlist)

if str == reversedstr:
  print("your string is a palindrome")
else:
  print("your string is not a palindrome")
```

```python
# 8. Input a text file (containing 1 or more paragraphs of English text) from the user,
# parse this file to display the frequency of occurrence of each word in this text file.
# Find the 3 most frequent words as well.
paragraph = input("please paste your paragraph below\n\n")
paragraph = paragraph.strip()
paragraphlist = paragraph.split(" ")

newlist = list(set(paragraphlist))

for word in newlist:
  print(f"element = {word} and count = {paragraphlist.count(word)}")
```
